```python
import numpy as np 
import tensorflow as tf


docs = [
    "hello world",
    "i like machine learning",
    "rnn models are useful for text",
    "deep learning is powerful",
    "natural language processing is fun",
    "this is a simple example",
    "training rnn requires sequences",
    "words can be represented as vectors",
    "sentences have variable length",
    "models learn from data"
]
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(oov_token='OOV')
```

```python
tokenizer.fit_on_texts(docs)
```

```python
tokenizer.word_index
```

```python
len(tokenizer.word_counts)
```

```python
tokenizer.document_count
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from tensorflow.keras.utils import pad_sequences
sequences = pad_sequences(sequences,padding='post')
sequences
```

### Embadding using keras 

```python
# before embading apply all the step that we apply before(up) for convert text to int(pad_sequence) Then....
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Embedding 

vocab_size = len(tokenizer.word_index) + 1
max_len = max(len(seq) for seq in tokenizer.texts_to_sequences(docs))

model = Sequential()
model.add(Embedding(vocab_size, output_dim=2, input_length=max_len))

model.summary()
```

```python
embeddings = model.predict(sequences)
print("Embeddings shape:", embeddings.shape)
print("First sequence embeddings:")
print(embeddings[0])
```
